```python
# QUESTION 1 : Basic Join

# Return:

# Customer name

# Account type

# Transaction amount

# Expected Output Example

# | name | account_type | amount |
```

```python
%%sql
SELECT 
    c.name,
    a.account_type,
    t.amount
FROM Customers c
JOIN Accounts a 
    ON c.customer_id = a.customer_id
JOIN Transactions t 
    ON a.account_id = t.account_id;
```

```python
# Explanation:

# customers → accounts join by customer_id

# accounts → transactions join by account_id

# This is a multi-table integration query.
```

```python
# QUESTION 2 — Total Transactions Per Customer

# Return:

# customer name

# total transaction amount

# Example Output

# | name | total_amount |
```

```python
%%sql
SELECT 
    c.name,
    SUM(t.amount) AS total_amount
FROM Customers c
JOIN Accounts a 
    ON c.customer_id = a.customer_id
JOIN Transactions t 
    ON a.account_id = t.account_id
GROUP BY c.name;
```

```python
# Key concept tested:

# GROUP BY

# SUM()
```

```python
# QUESTION 3 — Customers With More Than One Account

# Return:

# customer name

# number of accounts
```

```python
%%sql
SELECT 
    c.name,
    COUNT(a.account_id) AS total_accounts
FROM Customers c
JOIN Accounts a 
    ON c.customer_id = a.customer_id
GROUP BY c.name
HAVING COUNT(a.account_id) > 1;
```

```python
# Important concept:

# HAVING filters after aggregation
```

```python
# QUESTION 4 — Monthly Transaction Report

# Return:

# month

# total transaction amount

# Example

# | month | total |
```

```python
%%sql
SELECT 
    DATE_FORMAT(transaction_date, '%Y-%m') AS month,
    SUM(amount) AS total
FROM Transactions
GROUP BY month;
```

```python
# QUESTION 5 — Data Transformation (Important)

# Create a column called transaction_size:

# Rules:

# amount	        category
# < 50,000	         Small
# 50,000 – 80,000	Medium
# > 80,000	        Large

# Return:

# | transaction_id | amount | transaction_size |
```

```python
%%sql
SELECT 
    transaction_id,
    amount,
    CASE
        WHEN amount < 50000 THEN 'Small'
        WHEN amount BETWEEN 50000 AND 80000 THEN 'Medium'
        ELSE 'Large'
    END AS transaction_size
FROM Transactions;
```

```python
# Concept tested:

# CASE WHEN transformation

# This is common in data pipelines.
```

```python
# QUESTION 6 — Data Integration (Very Important)

# Create a dataset ready for reporting with:

# customer name

# city

# account_type

# total transactions

# Output example

# | name | city | account_type | total_amount |
```

```python
%%sql
SELECT 
    c.name,
    c.city,
    a.account_type,
    SUM(t.amount) AS total_amount
FROM Customers c
JOIN Accounts a 
    ON c.customer_id = a.customer_id
JOIN Transactions t 
    ON a.account_id = t.account_id
GROUP BY 
    c.name,
    c.city,
    a.account_type;
```

```python
# QUESTION 8 — Data Modeling

# Design a simple structure for loan management.

# You should suggest 3 tables and their relationships.

# Example format:

# Table 1
# Table 2
# Table 3


# Question 8 — Data Modeling (Loan System)
# 1️⃣ Customers Table

# Stores information about bank customers.

# column	                    type	            description
# customer_id	                INT (PK)        	unique customer ID
# name	VARCHAR	customer        name 
# phone	VARCHAR	customer        phone number

# Example:

# customer_id                	name	                phone
# 1	                            Asha	                0712345678
# 2	                            Kelvin              	0755555555

# 2️⃣ Loans Table

# Stores loans issued to customers.

# column	type	description
# loan_id	INT (PK)	unique loan ID
# customer_id	INT (FK)	references Customers
# loan_amount	DECIMAL	loan amount
# loan_date	DATE	date loan issued

# Example:

# loan_id	customer_id	loan_amount	loan_date
# 101	1	500000	2026-01-01
# 102	2	1000000	2026-02-10
# 3️⃣ Loan_Payments Table

# Stores repayments made for loans.

# column	type	description
# payment_id	INT (PK)	payment ID
# loan_id	INT (FK)	references Loans
# amount	DECIMAL	payment amount
# payment_date	DATE	payment date

# Example:

# payment_id	loan_id	amount	payment_date
# 1	101	100000	2026-02-01
# 2	101	50000	2026-03-01

# Relationships
# Customers
#    │
#    │ customer_id
#    ▼
# Loans
#    │
#    │ loan_id
#    ▼
# Loan_Payments

# Relationship explanation

# One customer → many loans

# One loan → many payments

# Why This Design is Good

# This structure is normalized because:

# Customer data is stored once

# Loan information is separated

# Payment records are stored independently

# Reduces data duplication

# Maintains referential integrity using foreign keys



# Optional (Very Strong Interview Answer)

# You can also say:

# "This design follows 3rd Normal Form (3NF) where each table stores a single concept and
# relationships are enforced using primary and foreign keys."
```

```python
# Bonus Question (Banks Love This)

# Return the top 2 customers with the highest total transactions.

# Expected output:

# | name | total_amount |
```

```python
%%sql
SELECT 
    c.name,
    SUM(t.amount) AS total_amount
FROM Customers c
JOIN Accounts a 
    ON c.customer_id = a.customer_id
JOIN Transactions t 
    ON a.account_id = t.account_id
GROUP BY c.name
ORDER BY total_amount DESC
LIMIT 2;
```
