```python
# QUESTION 1 : Basic Join

# Return:

# Customer name

# Account type

# Transaction amount

# Expected Output Example

# | name | account_type | amount |
```

```python
%%sql
SELECT 
    c.name,
    a.account_type,
    t.amount
FROM Customers c
JOIN Accounts a 
    ON c.customer_id = a.customer_id
JOIN Transactions t 
    ON a.account_id = t.account_id;
```

```python
# Explanation:

# customers → accounts join by customer_id

# accounts → transactions join by account_id

# This is a multi-table integration query.
```

```python
# QUESTION 2 — Total Transactions Per Customer

# Return:

# customer name

# total transaction amount

# Example Output

# | name | total_amount |
```

```python
%%sql
SELECT 
    c.name,
    SUM(t.amount) AS total_amount
FROM Customers c
JOIN Accounts a 
    ON c.customer_id = a.customer_id
JOIN Transactions t 
    ON a.account_id = t.account_id
GROUP BY c.name;
```

```python
# Key concept tested:

# GROUP BY

# SUM()
```

```python
# QUESTION 3 — Customers With More Than One Account

# Return:

# customer name

# number of accounts
```

```python
%%sql
SELECT 
    c.name,
    COUNT(a.account_id) AS total_accounts
FROM Customers c
JOIN Accounts a 
    ON c.customer_id = a.customer_id
GROUP BY c.name
HAVING COUNT(a.account_id) > 1;
```

```python
# Important concept:

# HAVING filters after aggregation
```

```python
# QUESTION 4 — Monthly Transaction Report

# Return:

# month

# total transaction amount

# Example

# | month | total |
```

```python
%%sql
SELECT 
    DATE_FORMAT(transaction_date, '%Y-%m') AS month,
    SUM(amount) AS total
FROM Transactions
GROUP BY month;
```

```python
# QUESTION 5 — Data Transformation (Important)

# Create a column called transaction_size:

# Rules:

# amount	        category
# < 50,000	         Small
# 50,000 – 80,000	Medium
# > 80,000	        Large

# Return:

# | transaction_id | amount | transaction_size |
```

```python
%%sql
SELECT 
    transaction_id,
    amount,
    CASE
        WHEN amount < 50000 THEN 'Small'
        WHEN amount BETWEEN 50000 AND 80000 THEN 'Medium'
        ELSE 'Large'
    END AS transaction_size
FROM Transactions;
```

```python
# Concept tested:

# CASE WHEN transformation

# This is common in data pipelines.
```

```python
# QUESTION 6 — Data Integration (Very Important)

# Create a dataset ready for reporting with:

# customer name

# city

# account_type

# total transactions

# Output example

# | name | city | account_type | total_amount |
```

```python
%%sql
SELECT 
    c.name,
    c.city,
    a.account_type,
    SUM(t.amount) AS total_amount
FROM Customers c
JOIN Accounts a 
    ON c.customer_id = a.customer_id
JOIN Transactions t 
    ON a.account_id = t.account_id
GROUP BY 
    c.name,
    c.city,
    a.account_type;
```

```python
# QUESTION 7 — Optimization Question

# You have this query:

# %%sql
# SELECT *
# FROM transactions
# WHERE account_id = 101;



# The table has 50 million rows and the query is slow.

# Question:
# What would you do to improve performance?
```

```python
# Best solution:

# Create an index.

# Why?

# Without index:Full Table Scan

# With index: Index Seek

# Huge improvement for millions of rows.

# Other good answers in interview:

# Avoid SELECT *

# Partition large tables

# Use proper indexing

# Filter early
```

```python
%%sql
CREATE INDEX idx_account_id
ON Transactions(account_id);
```

```python
# Bonus Question (Banks Love This)

# Return the top 2 customers with the highest total transactions.

# Expected output:

# | name | total_amount |
```

```python
%%sql
SELECT 
    c.name,
    SUM(t.amount) AS total_amount
FROM Customers c
JOIN Accounts a 
    ON c.customer_id = a.customer_id
JOIN Transactions t 
    ON a.account_id = t.account_id
GROUP BY c.name
ORDER BY total_amount DESC
LIMIT 2;
```
