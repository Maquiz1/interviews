```python
# Relationships (Very Important)
# Customers
#    │
#    │ 1-to-many
#    ↓
# Loans
#    │
#    │ 1-to-many
#    ↓
# Loan_Payments

# Meaning:

# One customer → many loans

# One loan → many payments


# ER Relationship Summary
# Table	Relationship
# Customers	1 → many Loans
# Loans	1 → many Loan_Payments
```

```python
# How to Explain It in the Interview

# You could say something like:

# "I would design three tables: Customers, Loans, and Loan_Payments.
# Each customer can have multiple loans, so the Loans table contains a foreign key referencing customer_id.
# Each loan can have multiple repayments, so Loan_Payments references loan_id.
# This structure keeps the data normalized and allows us to track loan balances and payment history efficiently."

# That explanation shows data modeling understanding.
```
